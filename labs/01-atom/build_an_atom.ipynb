# Lab 1.4 — Build an atom in code

**Goal:** write a function that takes numbers of protons, neutrons and electrons and tells you what you have built: the element, its isotope, its charge and its mass.

Rules from the lesson:

* the number of protons $Z$ decides the element
* the mass number is $A = Z + N$
* the charge is protons − electrons

```python
import sys; sys.path.append("..")
from constants import m_p, m_n, m_e, u, e as e_charge
import matplotlib.pyplot as plt
```

## 1. The first twenty elements

```python
SYMBOLS = ["", "H", "He", "Li", "Be", "B", "C", "N", "O", "F", "Ne",
           "Na", "Mg", "Al", "Si", "P", "S", "Cl", "Ar", "K", "Ca"]
NAMES = ["", "hydrogen", "helium", "lithium", "beryllium", "boron", "carbon", "nitrogen",
         "oxygen", "fluorine", "neon", "sodium", "magnesium", "aluminium", "silicon",
         "phosphorus", "sulfur", "chlorine", "argon", "potassium", "calcium"]

print(len(SYMBOLS) - 1, "elements, from", NAMES[1], "to", NAMES[20])
```

## 2. Describe an atom

Complete `describe(p, n, e)`. It returns a dictionary with the element, mass number, charge and notation such as `23Na+`.

```python
def describe(p, n, e):
    if not 1 <= p <= 20:
        raise ValueError("this lab knows elements with 1 to 20 protons")
    A = p + n
    charge = p - e
    if charge == 0:
        sign = ""
    else:
        sign = ("" if abs(charge) == 1 else str(abs(charge))) + ("+" if charge > 0 else "-")
    return {
        "element": NAMES[p],
        "symbol": SYMBOLS[p],
        "A": A,
        "charge": charge,
        "notation": f"{A}{SYMBOLS[p]}{sign}",
    }

for atom in [(1, 0, 1), (6, 6, 6), (6, 8, 6), (11, 12, 10), (17, 18, 18), (2, 2, 0)]:
    d = describe(*atom)
    print(f"{str(atom):14s} -> {d['notation']:7s} {d['element']:10s} charge {d['charge']:+d}")
```

## 3. How heavy is it?

Add up the particle masses. Then compare with the atomic mass unit, $u = 1.6605 \times 10^{-27}$ kg.

```python
def mass_kg(p, n, e):
    return p * m_p + n * m_n + e * m_e

for atom in [(1, 0, 1), (6, 6, 6), (8, 10, 8)]:
    m = mass_kg(*atom)
    print(f"{describe(*atom)['notation']:5s} {m:.4e} kg = {m / u:.3f} u")
```

Carbon-12 comes out at about 12.10 u, but by definition it is exactly 12 u. The missing 0.1 u is the **mass defect**: when protons and neutrons bind into a nucleus, a little mass turns into binding energy ($E = mc^2$, Module 6).

## 4. Where is the mass?

What fraction of each atom's mass sits in its nucleus?

```python
Z = list(range(1, 21))
# Use the most common isotope: roughly as many neutrons as protons (a little more for heavier atoms).
N = [0, 2, 4, 5, 6, 6, 7, 8, 10, 10, 12, 12, 14, 14, 16, 16, 18, 22, 20, 20]
fraction = [(p * m_p + n * m_n) / mass_kg(p, n, p) for p, n in zip(Z, N)]

plt.figure(figsize=(8, 3.5))
plt.plot(Z, [100 * f for f in fraction], "o-")
plt.xticks(Z, SYMBOLS[1:])
plt.ylabel("Mass in the nucleus (%)")
plt.title("Almost all of an atom's mass is in its nucleus")
plt.grid(alpha=0.3)
plt.show()
print(f"hydrogen: {100 * fraction[0]:.3f}%   calcium: {100 * fraction[-1]:.4f}%")
```

## 5. Charge in coulombs

Each unit of charge is $e = 1.602 \times 10^{-19}$ C.

```python
for atom in [(11, 12, 10), (8, 8, 10), (2, 2, 0)]:
    d = describe(*atom)
    print(f"{d['notation']:6s} charge {d['charge']:+d} e = {d['charge'] * e_charge:+.3e} C")
```
