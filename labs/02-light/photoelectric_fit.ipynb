# Lab 2.4 — Measure Planck's constant

**Goal:** use stopping-voltage measurements to find Planck's constant $h$ and identify an unknown metal, as Robert Millikan did in 1916.

Einstein's equation for the photoelectric effect is

$$KE_{max} = hf - \varphi$$

The stopping voltage $V_s$ turns back even the fastest electrons, so $eV_s = KE_{max}$. In electron-volts, a stopping voltage of 1.5 V means $KE_{max} = 1.5$ eV. Plot $KE_{max}$ against $f$: the slope is $h$ and the intercept is $-\varphi$.

```python
import sys; sys.path.append("..")
from constants import h, c, e
import numpy as np
import matplotlib.pyplot as plt
```

## 1. The measurements

A student shone five colours of light on an unknown metal and found the stopping voltage for each.

```python
wavelength_nm = np.array([220, 250, 300, 350, 400])
stopping_V    = np.array([2.75, 2.11, 1.24, 0.69, 0.21])

for lam, V in zip(wavelength_nm, stopping_V):
    print(f"{lam} nm  ->  stopping voltage {V:.2f} V")
```

## 2. Convert to frequency and energy

Complete the two lines below. Frequency is $f = c/\lambda$; in eV, $KE_{max}$ is numerically equal to $V_s$.

```python
f = c / (wavelength_nm * 1e-9)      # Hz
KE_eV = stopping_V * 1.0            # eV: e × V in joules, divided by e to get eV

print("f (×10¹⁴ Hz):", np.round(f / 1e14, 2))
print("KE_max (eV): ", KE_eV)
```

## 3. Fit a straight line

```python
slope, intercept = np.polyfit(f, KE_eV, 1)

h_eVs = slope                      # eV·s
h_Js = slope * e                   # J·s
phi = -intercept                   # eV
f0 = phi / slope                   # threshold frequency, Hz

print(f"Slope     h = {h_eVs:.3e} eV·s = {h_Js:.3e} J·s")
print(f"Accepted  h = {h / e:.3e} eV·s = {h:.3e} J·s")
print(f"Error       = {100 * (h_Js - h) / h:+.1f}%")
print(f"Work function φ = {phi:.2f} eV")
print(f"Threshold wavelength = {c / f0 * 1e9:.0f} nm")
```

## 4. Plot it

```python
f_line = np.linspace(0, 16e14, 100)

plt.figure(figsize=(7, 4.5))
plt.plot(f / 1e14, KE_eV, "o", label="measurements")
plt.plot(f_line / 1e14, slope * f_line + intercept, "-", label=f"fit: slope = {h_eVs:.2e} eV·s")
plt.axhline(0, color="gray", lw=0.8)
plt.xlabel("Light frequency (×10¹⁴ Hz)")
plt.ylabel("KE_max (eV)")
plt.title("Photoelectric effect: KE_max against frequency")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Which metal is it?

Compare your work function with these textbook values.

```python
metals = {"caesium": 2.10, "sodium": 2.28, "calcium": 2.87, "zinc": 4.30, "copper": 4.70}
best = min(metals, key=lambda m: abs(metals[m] - phi))
for m, w in metals.items():
    mark = "  <- closest" if m == best else ""
    print(f"{m:8s} {w:.2f} eV{mark}")
```

## Newton vs quantum

Wave theory says the electrons' energy should depend on the light's **brightness**. Your graph uses only **frequency**, and every point lands on one straight line. Brightness never appears in Einstein's equation.
